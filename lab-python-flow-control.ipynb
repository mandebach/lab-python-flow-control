```python
customer_orders = set()

keep_adding = "yes"
#Continue the loop until the user does not want to add another product.
while keep_adding == "yes":
# Prompt the user to enter the name of a product that a customer wants to order.
    product_name = input("Enter the name of the product you want to order: ")
# Add the product to the "customer_orders" set.
    customer_orders.add(product_name)
# Ask the user if they want to add another product (yes/no).
    keep_adding = input("Do you want to continue? (yes/no)")
```

```python
total_products_ordered = 0

for item in customer_orders:
    item_quantity =  int(input(f"How many {item} would you like to order? "))
    total_products_ordered = total_products_ordered + item_quantity
    
print("Your total products ordered are: ", total_products_ordered)
```
